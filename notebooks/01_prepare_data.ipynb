```python
import pandas as pd

# 1. Φόρτωση αρχικών δεδομένων
ct = pd.read_csv("../data/CTDAPD Dataset.csv")
ci = pd.read_csv("../data/cybersecurity_intrusion_data.csv")

# 2. Κρατάμε μόνο τις στήλες που ορίσαμε (είσοδοι + έξοδοι)
ct_cols = ["Destination_Port", "Protocol_Type", "Flow_Duration", "Packet_Size",
           "Total_Forward_Packets", "Total_Backward_Packets", "IDS_Alert_Count",
           "Anomaly_Score", "System_Patch_Status",                  # είσοδοι
           "Label", "Attack_Severity", "Attack_Vector"]             # έξοδοι

ci_cols = ["network_packet_size", "protocol_type", "login_attempts", "session_duration",
           "encryption_used", "ip_reputation_score", "failed_logins",
           "unusual_time_access",                                   # είσοδοι
           "attack_detected"]                                       # έξοδος

ct = ct[ct_cols].copy()
ci = ci[ci_cols].copy()

# 3. Η pandas διαβάζει τη λέξη "None" ως κενό (NaN) -> την επαναφέρουμε ως κατηγορία
ct["Attack_Vector"] = ct["Attack_Vector"].fillna("None")      # καμία γνωστή επίθεση
ci["encryption_used"] = ci["encryption_used"].fillna("None")  # χωρίς κρυπτογράφηση

# 4. Έλεγχος
print(ct.shape, ci.shape)
print("Κενές τιμές:", ct.isna().sum().sum(), ci.isna().sum().sum())
print(ct["Anomaly_Score"].describe())
print(ci["session_duration"].describe())

# 5. Αποθήκευση καθαρών αρχείων
ct.to_csv("../data/ctdapd_clean.csv", index=False)
ci.to_csv("../data/intrusion_clean.csv", index=False)
```
